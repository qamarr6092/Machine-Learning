```python
import sys
import os
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
sys.path.append('..')
```

```python
from src.preprocess import preprocess_data
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline
from sklearn.model_selection import learning_curve, GridSearchCV, StratifiedKFold, RandomizedSearchCV
from sklearn.metrics import confusion_matrix, classification_report, make_scorer, recall_score, auc, roc_curve, precision_recall_curve
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
```

```python
file_path = '../data/WA_Fn-UseC_-Telco-Customer-Churn.csv'
X_train, X_test, y_train, y_test, preprocesser = preprocess_data(df_filepath=file_path)

rf_clf = RandomForestClassifier(
    criterion='entropy', 
    bootstrap=True, 
    n_jobs=1
)

model = Pipeline([
    ('preprocessing', preprocesser),
    ('model', rf_clf)
])
```

```python
folds = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)
param_grid = {
    'model__n_estimators': [100, 200],
    'model__max_depth': [8, 16, None],
    'model__min_samples_leaf': [2, 5],
    'model__min_samples_split': [5, 10],
    'model__class_weight': ['balanced', 'balanced_subsample', None]
}
score = make_scorer(recall_score, pos_label=1)
random_search = RandomizedSearchCV(
    estimator=model,
    param_distributions=param_grid,
    n_iter=15,
    cv=folds,
    random_state=26,
    n_jobs=2,
    scoring=score
)

random_search.fit(X_train, y_train)
```

```python
svc = Pipeline([
    ('preprocessing', preprocesser),
    ('model', SVC(kernel='rbf', probability=True))
])

svc_param = {
    'model__gamma': [0.5, 5, 10, 25],
    'model__C' : [0.1, 2, 5, 20]
}
svc_cv = RandomizedSearchCV(
    estimator=svc, 
    param_distributions=svc_param,
    scoring=score, 
    cv=folds,
    n_iter=3
)

log = Pipeline([
    ('preprocessing', preprocesser),
    ('model', LogisticRegression())
])

log_param = {
    'model__C': [0.5, 15, 30]
}

log_cv = GridSearchCV(
    estimator=log, 
    param_grid=log_param,
    scoring=score, 
    cv=folds
)
log_cv.fit(X_train, y_train)
svc_cv.fit(X_train, y_train)
```
