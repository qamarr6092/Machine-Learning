```python
import xgboost
import sklearn
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns 
import joblib
import shap
```

```python
from xgboost import XGBRegressor, XGBClassifier
from sklearn.datasets import make_regression, load_diabetes, make_classification
from sklearn.metrics import root_mean_squared_error, r2_score
from sklearn.model_selection import train_test_split, learning_curve, GridSearchCV, LearningCurveDisplay
from sklearn.inspection import DecisionBoundaryDisplay
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import Pipeline
from matplotlib.colors import ListedColormap
from sklearn.ensemble import RandomForestClassifier
```

```python
X, y = make_regression(
    n_samples=100,
    n_features=1,
    n_informative=1, 
    random_state=26,
    coef=False, 
    noise=2
)

model = XGBRegressor(
    objective='reg:squarederror',
    eval_metric='rmse',
    n_estimators=5,
    learning_rate=0.3,
    max_depth=10,
    random_state=26,
    max_leaves=3
)
model.fit(X, y)
x = np.linspace(X.min(), X.max(), 150)

plt.figure()
plt.scatter(X, y)
plt.plot(x, model.predict(x), color='r', label='Xgboost predicition')
plt.legend()
plt.grid()
plt.show()
```

```python
df = load_diabetes(as_frame=True).frame
df.head()
```

```python
X = df.drop(columns='target')
y = df['target']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=26)

model = XGBRegressor(
    objective='reg:squarederror',
)

param_grid = {
    'n_estimators': [50, 150, 300],
    'max_depth': [3, 4, 6],          
    'learning_rate': [0.05, 0.1],  
    'gamma': [2, 5, 15],
    'reg_lambda': [0.1, 2, 5]
 
}

grid_search = GridSearchCV(
    model, param_grid,
    scoring='neg_root_mean_squared_error', 
    cv=5,
)

grid_search.fit(X_train, y_train)
```

```python
grid_search.best_score_, grid_search.best_params_
```

```python
model = grid_search.best_estimator_

data = pd.DataFrame({
    'Features': model.feature_names_in_, 
    'Importance': model.feature_importances_
}).sort_values(by='Importance', ascending=False)
plt.figure()
sns.barplot(data=data, x='Importance', y='Features')
plt.show()
```

```python
r2_score(y_train, model.predict(X_train)), r2_score(y_test, model.predict(X_test))
```

```python
model_2 = LinearRegression().fit(X_train, y_train)

r2_score(y_train, model_2.predict(X_train)), r2_score(y_test, model_2.predict(X_test))
```

```python
model_3 = Pipeline([
    ('poly', PolynomialFeatures(degree=2, include_bias=False)),
    ('model', LinearRegression())
])
model_3.fit(X_train, y_train)

r2_score(y_train, model_3.predict(X_train)), r2_score(y_test, model_3.predict(X_test))
```

```python
fig, ax = plt.subplots(1, 2, figsize=(14,6))
LearningCurveDisplay.from_estimator(
    model,
    X_train, 
    y_train, 
    cv=5, 
    scoring='r2',
    ax=ax[0],
    train_sizes=np.linspace(0.1, 1, 20)
)
ax[0].set_title('Xgboost')

LearningCurveDisplay.from_estimator(
    model_2,
    X_train, 
    y_train, 
    cv=5, 
    scoring='r2',
    ax=ax[1],
    train_sizes=np.linspace(0.1, 1, 20)
)
ax[1].set_title('Linear Regression')

plt.show()
```

```python
plt.scatter(y_test, model.predict(X_test))
plt.scatter(y_test, model_2.predict(X_test), color='r')
plt.show()
```

```python
model = XGBClassifier(
    objective='multi:softprob', 
    num_class=4,                 
    eval_metric='mlogloss',
    n_estimators=100,
    learning_rate=0.1,
    max_depth=6,
    random_state=26,
)

model_2 = RandomForestClassifier(
    criterion='entropy',
)
model_2.fit(X, y)

X, y = make_classification(
    n_samples=250,
    n_features=2, 
    n_informative=2, 
    n_redundant=0,
    n_classes=4,
    n_clusters_per_class=1,
    random_state=26
)

df = pd.DataFrame({
    'X1': X[:, 0],
    'X2': X[:, 1],
    'y' : y
})

model.fit(X, y)

fig, ax = plt.subplots(1, 2, figsize=(14,6))
display = DecisionBoundaryDisplay.from_estimator(
    model, X,
    alpha=0.7,
    plot_method='contourf', response_method='predict',
    ax=ax[0],
    cmap=ListedColormap(['#FF6B6B', '#4ECDC4', '#FFE66D', '#A8E6CF'])
)
ax[0].set_title('Xgboost')
sns.scatterplot(data=df, x='X1', y='X2', hue='y', palette='Set1', s=50, ax=ax[0])

display = DecisionBoundaryDisplay.from_estimator(
    model_2, X,
    alpha=0.7,
    plot_method='contourf', response_method='predict',
    ax=ax[1],
    cmap=ListedColormap(['#FF6B6B', '#4ECDC4', '#FFE66D', '#A8E6CF'])
)
ax[1].set_title('Random Forest')
sns.scatterplot(data=df, x='X1', y='X2', hue='y', palette='Set1', s=50, ax=ax[1])

plt.show()
```
