```python
from sklearn.datasets import load_breast_cancer
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns 

df = load_breast_cancer(as_frame =True).frame
df.head(4)
```

whether a breast tumor is malignant (cancerous) or benign (non-cancerous

```python
columns = df.columns 
df.describe()
```

```python
df['target'].value_counts()
```

```python
df[columns].hist(
    bins=20,
    figsize=(14, 12),
    grid=False,
    color='royalblue',
    edgecolor='black',
    linewidth=0.5,
    alpha=0.7
)

plt.suptitle('Distribution ofFeatures', fontsize=16)
plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(5, 2, figsize=(12, 15))
ax = ax.flatten()

for idx, val in enumerate(columns[:10]):
    sns.boxplot(
        data=df,
        x='target',
        y=val,
        ax=ax[idx],
        width=0.5,
        hue = 'target',
        palette='Set2',                
        linewidth=1.5              
    )
    ax[idx].set_title(f'{val} by Target', fontsize=10)
    ax[idx].grid(axis='y', alpha=0.6)

plt.suptitle('Feature Distribution by Target (0=Malignant, 1=Benign)', fontsize=14)
plt.tight_layout()
plt.show()
```

In the dataset there are bunch of features that are heavily correlated which reduces the model perfomance

```python
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.model_selection import train_test_split ,cross_val_predict,cross_val_score,StratifiedKFold

X = df.drop( 'target' , axis = 1) 
y = df['target']

X_train ,X_test , y_train ,y_test = train_test_split( X,y ,random_state = 26, test_size = 0.2 , shuffle = True , stratify=y)

preprocess = Pipeline ( [ ('scaling' , StandardScaler() ) ,
                          ('dimension_reduction' , PCA (n_components=0.98))
                        ])

model = LogisticRegression()
X_train_transformed = preprocess.fit_transform(X_train)

model.fit(X_train_transformed , y_train)
```
