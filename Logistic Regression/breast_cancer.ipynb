In the dataset there are bunch of features that are heavily correlated which reduces the model perfomance

```python
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.model_selection import train_test_split ,cross_val_predict,cross_val_score,StratifiedKFold ,GridSearchCV
from sklearn.metrics import confusion_matrix ,precision_score , recall_score , f1_score ,make_scorer , precision_recall_curve,roc_auc_score,roc_curve,auc

X = df.drop( 'target' , axis = 1) 
y = df['target']

X_train ,X_test , y_train ,y_test = train_test_split( X,y ,random_state = 26, test_size = 0.2 , shuffle = True , stratify=y)

pipeline = Pipeline ( [ ('scaling' , StandardScaler() ) ,
                          ('dimension_reduction' , PCA ()),
                          ('model' , LogisticRegression())
                        ])
skfolds = StratifiedKFold(n_splits = 5, shuffle=True)

param_grid = { 'dimension_reduction__n_components' : [0.85,0.90,0.95,0.98,0.99] }

recall = make_scorer( recall_score , pos_label = 0 ) 

grid_search = GridSearchCV(pipeline, param_grid, scoring = recall ,cv =skfolds)

grid_search.fit(X_train,y_train)
```

```python
pd.DataFrame(grid_search.cv_results_).sort_values(by= 'rank_test_score')
```

```python
grid_search.best_estimator_
```

```python
print(grid_search.best_estimator_)
```

```python
grid_search.best_params_
```

```python
grid_search.best_score_
```

```python
my_model = grid_search.best_estimator_
```

```python
y_train_pred = cross_val_predict( my_model ,X_train ,y_train ,cv = skfolds, method = 'predict')

labels = ['all' , 'true' ,'pred']

fig ,ax = plt.subplots( 2,2 , figsize = (12,8))
ax = ax.flatten()
plt.suptitle('Confusion Matrix (all ,recall, precision)')
for idx , val in enumerate ( labels ) :
    confusion = confusion_matrix( y_train ,y_train_pred, normalize = val )
    sns.heatmap (confusion , ax = ax[idx] ,fmt ='.2f' ,linecolor = 'black' , linewidth =1 , annot_kws = {'size' : 15} , annot =True, cmap ='Blues')

plt.show()
```

Normalize = 'pred' how many correct given the predicition = TP/TP+FP = precision

Normalize = 'true' how many correct given the actual = TP/TP+FN = recall

```python
f1_score(y_train, y_train_pred,pos_label = 0)
```

```python
y_train_proba = cross_val_predict(my_model,X_train,y_train,method='predict_proba',cv=skfolds)

precision_0, recall_0, thresholds_0 = precision_recall_curve(y_train, y_train_proba[:, 0], pos_label=0)
precision_1, recall_1, thresholds_1 = precision_recall_curve(y_train, y_train_proba[:, 1], pos_label=1)

fig, ax = plt.subplots(1, 2, figsize=(12, 6))

ax[0].plot(recall_0, precision_0, label='Class 0', color='blue')
ax[0].set_xlabel('Recall')
ax[0].set_ylabel('Precision')
ax[0].set_title('Precision-Recall Curve (Class 0)')
ax[0].grid(True, alpha=0.3)
ax[0].legend()

ax[1].plot(recall_1, precision_1, label='Class 1', color='red')
ax[1].set_xlabel('Recall')
ax[1].set_ylabel('Precision')
ax[1].set_title('Precision-Recall Curve (Class 1)')
ax[1].grid(True, alpha=0.3)
ax[1].legend()

plt.show()
```

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 6))

ax[0].plot(thresholds_0, precision_0[:-1], label='precsion', color='blue')
ax[0].plot(thresholds_0, recall_0[:-1], label='recall', color='red')
ax[0].set_xlabel('threshold')
ax[0].grid(True, alpha=0.3)
ax[0].set_title('(Class 0)')
ax[0].legend()

ax[1].plot(thresholds_1, precision_1[:-1], label='precsion', color='blue')
ax[1].plot(thresholds_1, recall_1[:-1], label='recall', color='red')
ax[1].set_xlabel('threshold')
ax[1].set_title('(Class 1)')
ax[1].grid(True, alpha=0.3)
ax[1].legend()

plt.show()
```

```python
fpr , tpr, thr = roc_curve(y_train ,y_train_proba[:,0],pos_label = 0 )

plt.figure(figsize=(8, 6))
plt.plot(fpr, tpr)
plt.plot([0, 1], [0, 1], 'k--')
plt.xlabel('False positive Rate (FPR)')
plt.ylabel('True positive Rate (TPR) recall')
plt.title('ROC Curve class 0')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
auc(fpr, tpr)
```

```python
youden_index = tpr - fpr
best_idx = np.argmax(youden_index)

best_threshold = thr[best_idx]
best_tpr = tpr[best_idx]
best_fpr = fpr[best_idx]


print(f"Best threshold: {best_threshold:.4f}")
print(f"TPR (Recall) at this threshold: {best_tpr:.4f}")
print(f"FPR at this threshold: {best_fpr:.4f}")
```

```python
y_train_proba_new_thr = (y_train_proba >= best_threshold).astype(int)



labels = ['all' , 'true' ,'pred']

fig ,ax = plt.subplots( 1,3 , figsize = (15, 5))
ax = ax.flatten()
plt.suptitle('Confusion Matrix (all ,recall, precision) New threshold')
for idx , val in enumerate ( labels ) :
    confusion = confusion_matrix( y_train ,y_train_proba_new_thr[:,1], normalize = val )
    sns.heatmap (confusion , ax = ax[idx] ,fmt ='.2f' ,linecolor = 'black' , linewidth =1 , annot_kws = {'size' : 15} , annot =True, cmap ='Blues')

plt.show()
```

```python
y_proba = my_model.predict_proba(X_test) 
y_pred = (y_proba >= best_threshold).astype(int)

labels = ['all' , 'true' ,'pred']

fig ,ax = plt.subplots( 1,3 , figsize = (15, 5))
ax = ax.flatten()
plt.suptitle('Confusion Matrix (all ,recall, precision) testing dataset')
for idx , val in enumerate ( labels ) :
    confusion = confusion_matrix( y_test ,y_pred[:,1], normalize = val )
    sns.heatmap (confusion , ax = ax[idx] ,fmt ='.2f' ,linecolor = 'black' , linewidth =1 , annot_kws = {'size' : 15} , annot =True, cmap ='Blues')

plt.show()
```
